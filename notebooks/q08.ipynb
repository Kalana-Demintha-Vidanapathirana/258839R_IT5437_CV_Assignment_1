## Q8 - Salt and Pepper Noise Removal


### 1 - Import Required Libraries


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

### 2 - Set Input and Output Paths


```python
img_path = Path("../data/raw/emma_corrupted.jpg")   # change to .jpg if needed
out_dir = Path("../results/q08")
out_dir.mkdir(parents=True, exist_ok=True)

print("Input path:", img_path.resolve())
print("Output folder:", out_dir.resolve())
```

### 3 - Load Corrupted Image

```python
img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)

if img is None:
    raise FileNotFoundError(f"Could not load image: {img_path}")

print("Image loaded successfully")
print("Shape:", img.shape)
print("Data type:", img.dtype)
```

### 4 - Display Corrupted Image

```python
plt.figure(figsize=(6, 6))
plt.imshow(img, cmap="gray")
plt.title("Corrupted Image")
plt.axis("off")
plt.show()
```

### 5 - Apply Gaussian Smoothing

```python
gaussian_smoothed = cv2.GaussianBlur(img, (5, 5), sigmaX=1.0, sigmaY=1.0)
print("Gaussian smoothing completed.")
```

### 6 - Apply Median Filtering

```python
median_filtered = cv2.medianBlur(img, 5)
print("Median filtering completed.")
```

### 7 - Display Filtering Results

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Corrupted Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(gaussian_smoothed, cmap="gray")
plt.title("Gaussian Smoothing")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(median_filtered, cmap="gray")
plt.title("Median Filtering")
plt.axis("off")

plt.tight_layout()
plt.show()
```

### 8 - Save Output Images

```python
cv2.imwrite(str(out_dir / "q08_original_corrupted.png"), img)
cv2.imwrite(str(out_dir / "q08_gaussian_smoothed.png"), gaussian_smoothed)
cv2.imwrite(str(out_dir / "q08_median_filtered.png"), median_filtered)

print(f"Saved corrupted image to: {out_dir / 'q08_original_corrupted.png'}")
print(f"Saved Gaussian result to: {out_dir / 'q08_gaussian_smoothed.png'}")
print(f"Saved median result to: {out_dir / 'q08_median_filtered.png'}")
```

## Interpretation and Discussion

The corrupted image contains salt-and-pepper noise, which appears as random bright and dark pixels.

Gaussian smoothing reduces the noise by averaging neighboring pixel values, but it also blurs edges and does not remove salt-and-pepper noise very effectively.

Median filtering replaces each pixel with the median value in its neighborhood. This makes it much more effective for removing salt-and-pepper noise while preserving edges better than Gaussian smoothing.

Therefore, for this type of noise, the median filter gives the better result.
